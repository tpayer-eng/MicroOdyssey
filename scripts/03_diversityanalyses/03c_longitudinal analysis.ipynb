```python
import os
import pandas as pd
import qiime2 as q2
from qiime2 import Visualization

%cd /home/jovyan/MicroOdyssey/scripts

data_dir = "../data"
results_dir = "../results/diversity"
core_dir = f"{results_dir}/core-metrics-results"
```

## 1. Changes in microbiome composition between consecutive visits

```python
#calculate bray–curtis distances
! qiime longitudinal first-distances \
    --i-distance-matrix $core_dir/bray_curtis_distance_matrix.qza \
    --m-metadata-file $data_dir/metadata_q2.tsv \
    --p-state-column host_age \
    --p-individual-id-column host_id \
    --o-first-distances $results_dir/bray_first_distances.qza
```

```python
! qiime longitudinal volatility \
    --m-metadata-file $data_dir/metadata_q2.tsv \
    --m-metadata-file $results_dir/bray_first_distances.qza \
    --p-state-column host_age \
    --p-individual-id-column host_id \
    --p-default-metric Distance \
    --p-default-column delivery_mode \
    --o-visualization $results_dir/bray_volatility.qzv
```
