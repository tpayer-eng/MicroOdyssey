## 1. Import datasets

```python
data_dir = "../data"
results_dir = "../results/diversity"

os.makedirs(results_dir, exist_ok=True)
```

```python
metadata = pd.read_csv(f"{data_dir}/metadata.csv", sep=";")
```

```python
#preparing metadata for QIIME 2 (tsv more suitable)
metadata_q2 = metadata.rename(columns={"sample_id": "sample-id"})
metadata_q2 = metadata_q2.set_index("sample-id")

metadata_q2.to_csv(f"{data_dir}/metadata_q2.tsv", sep="\t")
```

## 5. Statistical testing

```python
# age is "birth" or in "months", so we need to convert this into numerical values
# alpha-correlation needs numeric metadata for Spearman correlation.

metadata_q2["age_months"] = (metadata_q2["host_age"]
    .replace({"birth": "0"})
    .astype(str)
    .str.extract(r"(\d+)", expand=False)
    .astype(float)
)

metadata_q2.to_csv(f"{data_dir}/metadata_q2.tsv", sep="\t")
```

```python
##REMOVE THIS AT THE END
print(metadata_q2[["host_age", "age_months"]].head(10))
```

```python
! qiime diversity alpha-group-significance \
  --i-alpha-diversity $results_dir/core-metrics-results/shannon_vector.qza \
  --m-metadata-file $data_dir/metadata_q2.tsv \
  --o-visualization $results_dir/shannon-group-significance.qzv
```

```python
Visualization.load(f"{results_dir}/shannon-group-significance.qzv")
```

```python
! qiime diversity alpha-correlation \
  --i-alpha-diversity $results_dir/core-metrics-results/shannon_vector.qza \
  --m-metadata-file $data_dir/metadata_q2.tsv \
  --o-visualization $results_dir/shannon-correlation.qzv
```

```python
Visualization.load(f"{results_dir}/shannon-correlation.qzv")
```

## 6. Visualization

```python
# load Shannon diversity values 
shannon = q2.Artifact.load(f"{results_dir}/core-metrics-results/shannon_vector.qza").view(pd.Series)

shannon.name = "shannon"

# combine Shannon diversity w sample metadata
alpha_df = metadata_q2.join(shannon, how="inner")

alpha_df.head()
```

```python
plt.figure()

plt.scatter(
    alpha_df["age_months"],
    alpha_df["shannon"],
    alpha=0.6
)

plt.xlabel("infant age (months)")
plt.ylabel("shannon diversity")
plt.title("gut microbiome diversity across infant age")

plt.tight_layout()
plt.show()
```

```python
# DO BOXPLOT?
```

```python
## CHECK THIS ONE
plt.figure()

for host_id, group in alpha_df.groupby("host_id"):
    group = group.sort_values("age_months")

    plt.plot(
        group["age_months"],
        group["shannon"],
        marker="o",
        alpha=0.3
    )

plt.xlabel("infant age (months)")
plt.ylabel("shannon diversity")
plt.title("individual infant microbiome diversity trajectories")

plt.tight_layout()
plt.show()
```
