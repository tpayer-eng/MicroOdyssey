# 5. bias from one child having multiple measurements?

```python
X = q2.Artifact.load(f"{data_dir}/table-enriched.qza").view(pd.DataFrame)
meta = pd.read_csv(f"{data_dir}/metadata.tsv", sep="\t", index_col="sample-id")
meta = meta.loc[X.index]
y, groups = meta["host_age"], meta["host_id"]   

rf = RandomForestRegressor(n_estimators=500, random_state=0, n_jobs=-1)

# random (leaky) vs grouped CV, same model
from sklearn.model_selection import KFold
for name, cv in [("random KFold", KFold(5, shuffle=True, random_state=0)),
                 ("GroupKFold", GroupKFold(5))]:
    pred = cross_val_predict(rf, X, y, cv=cv, groups=groups if name == "GroupKFold" else None)
    print(f"{name}: MAE={mean_absolute_error(y, pred):.2f}, R2={r2_score(y, pred):.2f}")
```

#### --> NO bias (potentially "leaky" model and clean CV model yield very similar R2"
